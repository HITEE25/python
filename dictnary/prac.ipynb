```python
try:
    a=int(input("enter a number : "))
    b = int(input("enter another number : "))
    c=a/b
except ZeroDivisionError:
    print("division with zero is not possible ")
except ValueError:
    print("the types of the numbers are not integers")
else:
    print("result is ", c)
finally:
    print("the try except block is successfully exceuted ")
```

```python
try:
    lst = [1,2,3,4]
    dic = {1:'a' , 2:'b' , 3:'c'}
    # print(lst[100])
    print(dic[100])
except IndexError:
    print("index is out of range")
except KeyError:
    print("key of the dict is not valid")
except Exception as e:
    print(e)
else:
    print("working fine")
finally:
    print("try block successsfully executed ")
```

```python
try:
    filename = input("enter the file name : ")
    with open(filename , "r") as f:
        pass
except FileNotFoundError:
    print("the file not found")
except NameError:
    print("name error is encountered")
else:
    print("the file exist and working fine")
finally:
    print("the try block is successfully executed")
```

```python
try:
    a = int(input("enter first number"))
    b = int(input("enter the second number"))
    sub = a-b
    div = a/b
except ZeroDivisionError:
    print("cant divide by zero")
except ValueError:
    print("the data types are not compatible")
except TypeError:
    print("the data type are invalid")
else:
    print(sub)
    print(div)
finally:
    print("try block is over")
```

```python
marks = int(input("enter the marks : "))
if(marks>100 or marks<0):
    raise ValueError("the marks must be between 0 to 100")

if(marks>90):
    print("A")
elif marks>=80:
    print("B")
elif marks>=70:
    print("c")
else:
    print("Fail")
```

```python
def findLargest(lst):
    assert len(lst)!=0 , "the length of the list is zero"
    temp = tuple(lst)
    max_el = max(temp)
    print("max element is : " , max_el)
lst = [1,2,3,4,5] 
findLargest(lst)
```
