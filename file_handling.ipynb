## a program to count the no of word and lines in a file 

```python
with open("ab.txt" , "r") as f:
    lines = f.readlines()

no_of_lines = len(lines)
no_of_words = 0
for i in lines:
    no_of_words+=len(i.split())
print(no_of_lines)
print(no_of_words)
```

## Program to separate even and odd numbers into two files

```python
with open("nums.txt" , "r") as f:
    nums = f.read().split()

with open("fe.txt" , 'w') as fe , open("fo.txt" , 'w') as fo:
    for num in nums:
        if(int(num)%2==0):
            fe.write(num+' ')
        else:
            fo.write(num+' ')

```
