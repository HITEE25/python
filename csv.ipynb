## Python provides a built-in csv module to work with CSV files.

```python
import csv
```

```python
## using csv.reader

with open("students.csv" , "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

## Writing to a CSV File

```python
with open("students.csv" , "w" , newline="") as f:
    writer = csv.writer(f)
    writer.writerow([6 , 'hari' , 'C++' , 90])
    writer.writerow([7 , 'hariiii' , 'python' , 100])
    
```

```python
##Writing Multiple Rows

with open("students.csv" , "a" , newline="") as f:
    writer = csv.writer(f)
    writer.writerows([
        [8,"haraprasad" , "java" , 19],
        [9 , "duggu" , "python" , 50]
    ])
```
